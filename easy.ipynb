#1 Two sums:
Ex1:
Input: nums = [2,7,11,15], target = 9

Output: [0,1]

Explanation: Because nums[0] + nums[1] == 9, we return [0, 1]

Ex2:
Input: nums = [3,2,4], target = 6

Output: [1,2]

Ex3:

Input: nums = [3,3], target = 6

Output: [0,1]


```python
def twoSum(nums, target):
    myDict = {}
    for i in range(len(nums)):
        res = target - nums[i]
        if res in myDict:           
            return [i, myDict[res]] if i < myDict[res] else  [myDict[res], i]
        else:
            myDict[nums[i]] = i

twoSum([3, 3], 6)
```

#2 Palindrome Number:
Input: x = 121
Output: true

Explanation: 121 reads as 121 from left to right and from right to left.

Example 2:

Input: x = -121

Output: false

Explanation: From left to right, it reads -121. From right to left, it becomes 121-. Therefore it is not a palindrome.

Example 3:

Input: x = 10

Output: false

Explanation: Reads 01 from right to left. Therefore it is not a palindrome.

```python
#solution
#1. reverse string
#2. while loop

def palindrome(num):
    return str(num) == str(num)[::-1]

# palindrome(10)
```

```python
#2
def palindrome(num):
    i=0
    st = str(num)
    j=len(st)-1

    while i<=j:
        if st[i] != st[j]:
            return False
        else:
            i+=1
            j-=1
    return True
    
# palindrome(-121)
```

```python
#13 Roman to Integer
# reverse the string, if current < prev value, total subtract it, else add it -> to avoid next number out of scope issue

def romanToInt(s: str) -> int:
        rev = s[::-1]

        hash = {
          'I':1,
          'V':5,
          'X':10,
          'L':50,
          'C':100,
          'D': 500,
          'M': 1000 
        }
        output = 0
        prev = 0

        for x in rev:
            current = hash[x]
            if current < prev:
                output -= current
            else:
                output += current

            prev = hash[x]

        return output
romanToInt('MCMXCIV')
```

```python
#14 Longest Common Prefix
strs = ["flower", "flow", "flight"]
##strs = ["dog","racecar","car"]

#2 ways to solve this problem
def commonPrefix(strs:List[str])->str:
    if not strs: 
        return ''
        
    firstword = strs[0]
    for i in range(len(firstword)):
        for s in strs[1:]:
            if i>= len(s) or s[i] != firstword[i]:
                return firstword[:i]
    return firstword

commonPrefix(strs)
#Time complexity(n*m) n - firstword length, m - all words except 1st word in the list
#Space complexity O(1)
```

```python
#20 Valid Parentheses - little hard

# s = "()[]{}"
# s = "()"
s='()[]{}'
def validParentheses(str):
    hash= {
        ')': '(',
        '}': '{',
        ']': '['
    }

    stack = []

    for char in str:
        if char in hash:
            if not stack or stack[-1] != hash[char]:
                return False
            stack.pop()
        else:
            stack.append(char)

    return len(stack) == 0

validParentheses(s)
#Time O(n)
#Space O(n)
```

```python
#26 Remove Duplicates from Sorted Array
# Way1 For Loop
# Way2 Set
list = [0,0,1,1,1,2,2,3,3,4]
def removeDuplicate(list):
    if not list:
        return 0
    j = 0
    for i in range(1, len(list)):
        if list[i] != list[j]:
            list[j+1] = list[i]
            j+=1
    return j+1

print(removeDuplicate(list))
```

```python
#Top K element frequent element
#way 1 use dict
#Time - nlogn sorting
#Space - O(n)
# Input: nums = [1,1,1,2,2,3], k = 2 -> Output: [1,2]
# nums = [1], k = 1
nums = [1,2,1,2,1,2,3,1,3,2]
k=2
def topKElement(nums, k):
    mapping = {}
    for x in nums:
        if x in mapping:
            mapping[x] +=1
        else:
            mapping[x] = 1

    return [key for key, _ in sorted(mapping.items(), key=lambda x: x[1], reverse = True)][:k]
print(topKElement(nums, 2))
#Time complexity - O(nlogn)
#Space Complexity - O(n)
```

```python
#Way 2
#using heap
import heapq

def top_k_keys(d, k):
    return [key for key, _ in heapq.nlargest(k, d.items(), key=lambda x: x[1])]

print(top_k_keys({1:3, 2:2, 3:1}, 2))  # [1, 2]
#Time complexity - nlogK better than nlogn
#Space O(n)
```

```python
#Way 3
nums = [1,1,1,2,2,3,3,3,3]
k = 2
from collections import Counter

def top_k_keys(nums, k):
    count = Counter(nums) # return a dict

    buckets = [[] for x in range(len(nums)+1)]

    for key, value in count.items():
        buckets[value].append(key)

    res = []

    for i in range(len(buckets)-1, -1, -1):
        for num in buckets[i]:
            res.append(num)
            if len(res) == k:
                return res
    return res
print(top_k_keys(nums, k))
#Time complexity is O(n) not O(n^2) -> each element only visit once
```

```python
#21 Merge Two Sorted Lists
# Input: list1 = [1,2,4], list2 = [1,3,4]
# Output: [1,1,2,3,4,4]

#Link List Question - Be familiar with linklist first to solve this problem: 

list1 = [1,2,4]
list2 = [1,3,4]
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        if not list1:
            return list2
        if not list2:
            return list1
        
        if list1.val < list2.val:
            list1.next = self.mergeTwoLists(list1.next, list2)
            return list1
        
        else:
            list2.next = self.mergeTwoLists(list2.next, list1)
            return list2

def build_list(arr):
    dummy = ListNode(0)
    cur = dummy
    for num in arr:
        cur.next = ListNode(num)
        cur = cur.next
    return dummy.next


def print_list(node):
    res = []
    while node:
        res.append(node.val)
        node = node.next
    print(res)


list1 = build_list([1,2,4])
list2 = build_list([1,3,4])
s = ListNode()
result = s.mergeTwoLists(list1, list2)

print_list(result)
```
